**Phase 4: Data Manipulation & Cleaning**

Task 9 : Fill missing Age values with the median.

```python
median_age = df["Age"].median()
df["Age"] = df["Age"].fillna(median_age)
```

```python
# Verify that there are no missing Age values
print("Median Age:", median_age)
print("Missing Age values after cleaning:", df["Age"].isnull().sum())
```

Task 10: Filter surviving passengers (Survived == 1).

```python
survived_passengers = df[df["Survived"] == 1]
```

Task 11: Convert Fare column to a NumPy array.

```python
fare_array = np.array(df["Fare"])
```

Task 12: Calculate maximum and average Fare using np.max() and np.mean().

```python
max_fare = np.max(fare_array)
average_fare = np.mean(fare_array)

print("Maximum Fare:", max_fare)
print("Average Fare:", average_fare)
```

**Phase 5: Data Visualization**

Task 13: Plot histogram of Age.

```python
plt.hist(df["Age"], bins=10)
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.title("Distribution of Passenger Age")
plt.show()
```

Task 14: Plot bar chart of Survival Count.

```python
survival_count = df["Survived"].value_counts().sort_index()

plt.bar(["Did Not Survive", "Survived"], survival_count)
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.title("Survival Count")
plt.show()
```

Task 15 : Plot scatter graph of Age vs Fare

```python
plt.scatter(df["Age"], df["Fare"])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.show()
```
